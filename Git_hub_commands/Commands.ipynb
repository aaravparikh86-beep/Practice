to publish new branch
git push --set upstream origin (branch_name)

to merge a branch
git merge (branch_name)

to pull all the data from git hub
git fetch origin
git pull origin

to see branch status
git status

to see remote branches
git branch -r

to see all the branches
git branch -a

to see local branch
git branch

to delete branch
git branch -d (branch_name)

to switch to any branch
git checkout (branch_name)
git switch (branch_name)

to make a new branch
git checkout -b (branch name)

to upload file to git hub
git push

to upload file to git
git commit -m "comment"

to add file into staging enviornment
git add .

to clone reposetry
git clone (repositry url)

to change the directry(folder)
cd (direcry name)

To move to parent folder
cd..

to initialize git
git init
